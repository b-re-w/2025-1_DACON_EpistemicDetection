## Load Datasets

````python
DATA_ROOT = "./data"

train_dataset = SWUnivDaconDataset(DATA_ROOT, train=True, valid_ratio=0)

print(f"INFO: Dataset loaded successfully. Train - {len(train_dataset)}")
````

````python
type_list = ["colloquial", "formal", "mixed", "storytelling"]
chosen = choices(type_list, weights=[0.3, 0.3, 0.2, 0.1], k=len(train_dataset))

Counter(chosen)
````

````python
if os.path.exists(augmented_file):
    augmented_df = pd.read_csv(augmented_file)
    print(f"INFO: Augmented dataset loaded from {augmented_file}")
else:
    print(f"INFO: No augmented dataset found at {augmented_file}, creating new one.")
    augmented_df = train_dataset.raw.copy()
    augmented_df['paraphrased_text'] = [""] * len(augmented_df)
    augmented_df['type'] = chosen
    augmented_df = augmented_df.reindex(columns=["title", "paraphrased_text", "full_text", "type", "generated"])
    augmented_df.to_csv(augmented_file, index=False)
augmented_df
````

## Define Model

````python
model_id = "Qwen/Qwen3-32B-GGUF"
````

````python
model_id = "Qwen/Qwen3-14B-GGUF"
````

````python
model = Llama.from_pretrained(repo_id=model_id, filename="*Q8_0.gguf", n_ctx=40_960, verbose=False)
````

````python
user_prompt = lambda s, p: f"""/nothink

**ROLE**: You are a professional Korean text augmentation AI specialized in creating diverse AI-generated training data. Your task is to rewrite Korean texts (primarily Wikipedia articles) to generate realistic AI-style variations while maintaining the original meaning and structure.

## **CONTEXT AWARENESS**
- **Target Domain**: Korean Wikipedia articles (2022 dump) with occasional essays and news articles
- **Training Purpose**: Generate AI-labeled data to balance human vs AI text classification
- **Language Focus**: Korean language patterns and expressions
- **Output Usage**: Training data for AI text detection models

## Core Principles

### 1. Meaning and Structure Preservation
- **Completely maintain** the core topic, arguments, and factual information
- **Preserve all** proper nouns, dates, numbers, and technical terms exactly
- **Keep** the logical flow and argumentative structure intact
- **Maintain** the encyclopedic/informative tone typical of Wikipedia

### 2. AI-Style Expression Patterns
- **Slightly formal and structured** writing style characteristic of AI
- **More systematic** use of transitional phrases and conjunctions
- **Consistent** paragraph organization and formatting
- **Balanced** sentence length distribution (avoid too many short sentences)
- **Subtle repetition** of key concepts for emphasis

### 3. Korean Language Specifics & Style Variations
- **Style Selection** (choose one per text):
  - **colloquial**: 반말 위주 (-해, -야, -지, -거든, -잖아)
  - **formal**: 격식체 (-다/-는다/-이다, -했다/-한다)
  - **mixed**: 구어체 기본 + 문어체 한 문장씩 섞음
  - **storytelling**: 할머니가 아이에게 이야기해주는 따뜻한 말투 (-단다, -이란다, -했단다, -네, -지요)
- **Apply** different honorific levels: 해라체/하다체/해요체/반말체/이야기체
- **Vary** Korean conjunctive expressions:
  - colloquial: 그런데/근데/그래서/아무튼/어쨌든
  - formal: 그러므로/따라서/또한/한편/즉
  - storytelling: 그런데 말이야/그래서 말이지/옛날에는 말이야/그러니까 말야
- **Use** style-appropriate discourse markers:
  - colloquial: 사실/진짜/정말/막/좀
  - formal: 즉/특히/실제로/또한/한편
  - storytelling: 아가야/얘야/그런데 말이야/참 신기하지/알겠지

## Working Procedure

### Step 1: Text Analysis
- Identify the main topic and structure of the provided text
- Note key information that must be preserved (facts, names, dates)
- Analyze current content type (Wikipedia, essay, news, etc.)
- **Accept the user-specified style choice** (do not judge appropriateness)

### Step 2: AI-Style Adaptation Planning
- **Use the specified text style** (colloquial/formal/mixed/storytelling) regardless of original content
- Plan systematic vocabulary substitutions matching the specified style
- Design sentence structure variations suitable for the target style
- Prepare transitional expressions matching the specified register
- Balance between natural variation and AI-generation patterns

### Step 3: Implementation with Korean Focus
- Apply changes paragraph by paragraph
- **Validate each sentence** for Korean grammar rules before proceeding
- Verify natural flow in Korean context
- Check that the result sounds like well-written AI-generated Korean text
- **Final grammar review**: Check all particles, endings, and conjugations

## Modification Techniques for Korean

### Vocabulary Level (Style-Specific)
- **colloquial**: 중요해 → 핵심이야 → 진짜 중요해, 생각해 → 보면 → 느낌이야
- **formal**: 중요한 → 핵심적인 → 주요한, 발전 → 성장 → 진보
- **mixed**: 중요해. 이는 핵심적인 요소다. → 정말 중요한 거야. 이것은 주요한 특징이다.
- **storytelling**: 중요하단다 → 소중한 거야 → 참 귀한 거란다, 알겠지 → 이해하겠니 → 신기하지
- **Tone matching**: 사용하다(formal) ↔ 쓰다(colloquial) ↔ 써보는 거야(storytelling)

### Korean Sentence Patterns (by Style)
- **colloquial endings**: -해/-야/-지/-거든/-잖아/-는데/-니까
- **formal endings**: -했다/-한 것으로 나타났다/-하는 것으로 알려져 있다
- **mixed example**: "진짜 대박이야. 이는 중요한 발견이라고 할 수 있다."
- **storytelling endings**: -단다/-이란다/-했단다/-네/-지요/-거란다/-느니라
- **storytelling patterns**: "옛날에는 말이야..." / "그런데 아가야..." / "참 신기하지 않니?"

### Korean Discourse Markers (Style-Specific)
- **colloquial**: 그런데/근데/그래서/아무튼/어쨌든/막/좀/진짜/사실
- **formal**: 먼저/다음으로/마지막으로/결론적으로/즉/다시 말해
- **mixed transitions**: "그런데 말이야" + "이를 통해 알 수 있듯이"
- **storytelling**: 그런데 말이야/그래서 말이지/옛날에는/그러니까/아가야/얘야/참말로

## Korean-Specific Precautions

### Elements to Preserve Exactly
- **Sino-Korean technical terms** (한자어 전문용어)
- **Korean proper nouns** (고유명사)
- **Numbers and dates** (수치 및 날짜)
- **Direct quotes** (인용문)
- **Official names and titles** (공식 명칭)

### Avoid These Changes
- Mixing different formality levels inappropriately **within the same chosen style**
- Using unnatural Korean expressions or awkward translations
- **In mixed style**: Imbalanced ratio between colloquial and formal (leaning too much to one side)
- **In storytelling style**: Using too formal academic language or overly childish expressions
- Creating grammatically incorrect Korean constructions
- **In colloquial style**: Using overly formal expressions
- **In formal style**: Including too casual expressions

## Output Format
- **Rewriting Result**: Complete modified Korean document in specified style with json format
  - Example: ```json
    {{ 'result': REWRITING RESULT }}
  ```

## Quality Standards for Korean AI Text
- **Expression change rate**: 60-80% expression change from original
- **Meaning accuracy**: 95%+ meaning preservation
- **Natural Korean**: Natural Korean grammar and flow
- **AI style consistency**: Consistent AI-like writing patterns in Korean
- **Style appropriateness**: Maintain chosen style (colloquial/formal/mixed) throughout

## AI Writing Style Characteristics to Emulate
- **Systematic information presentation**: Organized and structured delivery
- **Consistent writing style**: Uniform tone throughout the text
- **Strategic repetition**: Appropriate repetition of key concepts
- **Clear structure**: Well-organized paragraphs and sections
- **Objective tone**: Informative and neutral approach

---

**INPUT**:
  **STYLE**: {s}
  **TEXT**: {p}"""
````

````python
response_format = dict(
    type="json_object",
    schema=dict(
        type="object",
        properties=dict(
            result=dict(type="string")
        ),
        required=["result"]
    )
)

generation_params = dict(
    response_format=response_format,
    temperature=0.7,
    top_p=0.8,
    top_k=20,
    repeat_penalty=1.5,
    stream=True
)
````

````python
outputs = [model.create_chat_completion(
    messages = [dict(role="user", content=user_prompt(t, augmented_df.iloc[0]['full_text']))],
    **generation_params
) for t in type_list]
````

````python
# colloquial
for chunk in outputs[0]:
    delta = chunk['choices'][0]['delta']
    print(delta.get("content", ""), end="", flush=True)
````

````python
# formal
for chunk in outputs[1]:
    delta = chunk['choices'][0]['delta']
    print(delta.get("content", ""), end="", flush=True)
````

````python
# mixed
for chunk in outputs[2]:
    delta = chunk['choices'][0]['delta']
    print(delta.get("content", ""), end="", flush=True)
````

````python
# storytelling
for chunk in outputs[3]:
    delta = chunk['choices'][0]['delta']
    print(delta.get("content", ""), end="", flush=True)
````

## Iteration Loop

````python
FILE_PART = 0  # 0~9

PORTIONS = [
    [10000*0, 10000*1],
    [10000*1, 10000*2],
    [10000*2, 10000*3],
    [10000*3, 10000*4],
    [10000*4, 10000*5],
    [10000*5, 10000*6],
    [10000*6, 10000*7],
    [10000*7, 10000*8],
    [10000*8, 10000*9],
    [10000*9, len(train_dataset)]  # 97172
]
````

````python
start, end = PORTIONS[FILE_PART]
````

````python
print(f"Starting Augmentation from index {start} to {end}...")
for i in tqdm(range(start, end), desc=f"Augmenting {FILE_PART}"):
    output = model.create_chat_completion(
        messages = [dict(role="user", content=user_prompt(augmented_df.iloc[i]['type'], augmented_df.iloc[i]['full_text']))],
        **generation_params
    )

    paraphrased_text = ""
    for chunk in output:
        delta = chunk['choices'][0]['delta']
        if "content" in delta:
            paraphrased_text += delta['content']
            print(delta['content'], end="", flush=True)

    augmented_df.iloc[i]['paraphrased_text'] = loads(paraphrased_text)['result']
    augmented_df.to_csv(augmented_file.replace(".csv", "") + f"_{FILE_PART}.csv", index=False)
    start = i + 1  # record progress
````

## Combine Results

````python
total_augmented_df = pd.read_csv(augmented_file.replace(".csv", "") + "_0.csv")

for i, (start, end) in enumerate(PORTIONS[1:]):
    current = pd.read_csv(augmented_file.replace(".csv", "") + f"_{i}.csv")
    total_augmented_df.iloc[start:end] = current.iloc[start:end]

total_augmented_df.to_csv(augmented_file.replace(".csv", "") + ".csv", index=False)
total_augmented_df
````
